# 02 · Source model

The Mo tube emits a **bremsstrahlung continuum** (Kramers' law, cut off at the Duane–Hunt limit
$\lambda_\min = hc/eU$) plus the **characteristic K lines**, which appear only above the Mo K edge
(≈ 20.0 kV). Intensities here are in relative units; the absolute scale is set in Phase 3.

```
import matplotlib.pyplot as plt
import numpy as np

from braggsim.constants import MO_K_EDGE_KEV
from braggsim.source import continuum, lambda_min_pm, line_rates

plt.rcParams.update(
    {
        "axes.grid": True,
        "grid.color": "#e5e5e5",
        "lines.linewidth": 1.5,
        "axes.spines.top": False,
        "axes.spines.right": False,
    }
)
RAMP = ["#86b6ef", "#3987e5", "#1c5cab", "#104281"]  # one hue, light → dark = low → high U
```

## Continuum vs tube voltage

The short-wavelength edge moves as $\lambda_\min \approx 1239.84/U$ pm; Kramers' photon spectrum
peaks at $2\lambda_\min$.

```
lam = np.linspace(20, 250, 2000)
fig, ax = plt.subplots(figsize=(7, 4.2))
for U, color in zip((20, 25, 30, 35), RAMP, strict=True):
    ax.plot(lam, continuum(lam, U, 1.0), color=color, label=f"U = {U} kV")
    ax.axvline(lambda_min_pm(U), color=color, lw=0.8, ls=":")
ax.set(xlabel="wavelength λ (pm)", ylabel="continuum photons per pm (rel.)")
ax.legend(frameon=False)
plt.show()
```

## Characteristic lines vs tube voltage

$I_K \propto I\,(U/U_K - 1)^m$ with $m = 1.67$ (literature): zero up to the K edge, then a steep rise.
Kβ follows Kα at the fixed emission ratio.

```
U = np.linspace(0, 35, 351)
ka, kb = np.array([line_rates(u, 1.0, 1.0) for u in U]).T
fig, ax = plt.subplots(figsize=(7, 4.2))
ax.plot(U, ka, color="#2a78d6", label="Kα")
ax.plot(U, kb, color="#eb6834", label="Kβ")
ax.axvline(MO_K_EDGE_KEV, color="#8a8a8a", lw=0.8, ls=":")
ax.set(xlabel="tube voltage U (kV)", ylabel="line photons (rel.)")
ax.legend(frameon=False)
plt.show()
```
