# 01 · Bragg geometry

Bragg's law for the NaCl (200) planes, $n\lambda = 2d\sin\theta$ with $d = 282.01$ pm (LD P6.3.3.1).
A wavelength is reflected in order $n$ only while $n\lambda \le 2d$.

```
import matplotlib.pyplot as plt
import numpy as np

from braggsim.constants import D_NACL_PM, MO_KA_PM, MO_KB_PM
from braggsim.crystal import theta_from_lambda

plt.rcParams.update(
    {
        "axes.prop_cycle": plt.cycler(color=["#2a78d6", "#eb6834", "#1baf7a"]),
        "axes.grid": True,
        "grid.color": "#e5e5e5",
        "lines.linewidth": 1.5,
        "axes.spines.top": False,
        "axes.spines.right": False,
    }
)
```

## Expected glancing angles (leaflet Table 2)

```
for n in (1, 2, 3):
    ka, kb = theta_from_lambda(MO_KA_PM, n), theta_from_lambda(MO_KB_PM, n)
    print(f"n={n}:  θ(Kα) = {ka:6.2f}°   θ(Kβ) = {kb:6.2f}°")
```

## θ(λ) per order

Each order's curve stops at $\lambda = 2d/n$ (θ = 90°): beyond it there is no reflection.
Dots mark Mo Kα and Kβ.

```
lam = np.linspace(20, 2 * D_NACL_PM, 2000)
fig, ax = plt.subplots(figsize=(7, 4.2))
for n in (1, 2, 3):
    (line,) = ax.plot(lam, theta_from_lambda(lam, n), label=f"n = {n}")
    for lk in (MO_KA_PM, MO_KB_PM):
        ax.plot(lk, theta_from_lambda(lk, n), "o", ms=4, color=line.get_color())
ax.set(xlabel="wavelength λ (pm)", ylabel="glancing angle θ (°)", ylim=(0, 92))
ax.legend(frameon=False)
plt.show()
```
